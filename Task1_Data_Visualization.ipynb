```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob


file_path = glob.glob("/content/API_SP.POP.TOTL_DS2_en_csv*")[0]

df = pd.read_csv(file_path, skiprows=4)

df.head()
```

```python

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum().head(10))
```

```python

data = df[['Country Name', '2024']].copy()

data = data.dropna()

countries = ['India', 'China', 'United States', 'Indonesia',
             'Pakistan', 'Nigeria', 'Brazil', 'Bangladesh',
             'Russia', 'Japan']

data = data[data['Country Name'].isin(countries)]

data = data.sort_values(by='2024', ascending=False)

print(data)
```

```python

plt.figure(figsize=(10, 5))

plt.bar(data['Country Name'], data['2024'])

plt.title('Population of 10 Countries in 2024')
plt.xlabel('Countries')
plt.ylabel('Population')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

population = df['2024'].dropna()

plt.figure(figsize=(8, 5))
plt.hist(population, bins=20, edgecolor='black')

plt.title('Population Distribution in 2024')
plt.xlabel('Population')
plt.ylabel('Number of Countries and Regions')

plt.show()

```

### Results and Observations

1. India has the highest population among the 10 selected countries in 2024.
2. China has the second-highest population.
3. The bar chart makes it easy to compare population across countries.
4. The histogram shows the distribution of population values.
5. Population values vary significantly across countries and regions.

### Conclusion

Data visualization helps us understand population patterns and compare different countries. Bar charts and histograms make large datasets easier to interpret.
